# ML model for music dataset

```python
# Import statement for ML - Decision Tree algorithm
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import joblib
```

```python
music = pd.read_csv("../Data/music.csv")
music.head(3)
```

```python
music.shape
```

```python
music.describe()
```

Idea is to predict the genre of music for any given age & gender.

```python
# Create input dataset (X) - by convention input dataset is marked by 'X'
X = music.drop(columns = ['genre'])
X.head(5)
```

```python
# Create output dataset (y) - - by convention output dataset is marked by 'y'
y = music['genre']
y.head(5)
```

```python
# Generate training and testing dataset using music dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)

# train_test_split() takes input(X), output/predictions(y) and a test size value (20%, 30%, etc)
```

```python
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
```

```python
# Persist model & load models
# Store trained model (to a file)
joblib.dump(model, 'music_recommender.joblib')

# This creates a model file and saves it to the current directory.
# Next time when you run, you only need to run from this step.
```

```python
# Load the model
model = joblib.load('music_recommender.joblib')
```
